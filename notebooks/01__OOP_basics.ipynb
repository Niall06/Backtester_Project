COMPOSITION AND INHERITANCE

```python
#Inheritance:
#Represents an IS-A relationship
#ie A dog is an animal
#So in this case, a dog will inherit the class animal when written in code. 


```

```python
#Composition:
#Represents a HAS-A relationship
#Car has an engine
```

```python
#Example of wrong use of Inheritance:
```

```python
class Engine:
    def start(self):
        print('Engine started.')
#This is a simple class definfition for an engine. 

class Car(Engine):
    pass
#This is a wrong definition as the car is not a subclass of an engine. 
```

Correct use with composition

```python
class Engine:
    def start(self):
        print('Engine started.')

class Car:
    def __init__(self):
        self.engine = Engine()

#This means that car has an instance of engine, which is correct. 
```

Using Composition

```python
car = Car()
car.engine.start() # Output: Engine started.

```

When to use inheritance:
- True hierachal relationship
- Shared Behavior
- Clear Parent-Child Structure

When to use Composition:
- One class uses another 
- Flexible design needed 
- Avoid Tight Coupling

Python abstract base classes (abc module)

Abstract class:
A class that cannot be instantiated on its own; meant to be subclassed
They contain abstract methods, which are declared bu have no implementation
Abstract classes benefits:
1. Prevents instantion of the class itself
2. Requires children to use inherited abstract methods. 

```python
from abc import ABC, abstractmethod
class Vehicle(ABC):
    #to declare a class as abstract, we need to inherit from the ABC class.
    @abstractmethod
    def go(self):
        pass
    @abstractmethod
    def stop(self):
        pass
class Car(Vehicle):
    def go(self):
        print("Car is moving")
    def stop(self):
        print("Car has stopped")
```

```python
#These are abstract methods that must be implemented by any subclass of Vehicle.
#ABC stands for Abstract Base Class. It is a class that cannot be instantiated and is meant to be subclassed. The abstract methods defined in the ABC must be implemented by any concrete subclass.
```
