Stationary vs Non-Stationary 

A time series is strictly stationary if its mean and variance are constant over time and the covariance only depends only on the lag, not the time itself
It also usually means that there is no seasonality within the time series


For asset prices, raw stock prices (Pt) follow non-stationary random walk / geometric Browian motin (I(1)). 

Correlation vs Cointegration

Correlation - Measures the degree to whcih two assets move in the same direction at the same time $\rho_{X, Y} = \frac{\text{Cov}(X,Y)}{\sigma_X \sigma_Y}$)
Cointegration - Two non-stationary series Xt and Yt are coitnegrated if a spcific linear combination exists so that the resulting spread is staionary

Engle-Granger Two-Step & ADF Testing

Verifying that residual spread is genuinley mean reverting rather than spurios

1) Run OLS to extract the residual et
2) Run augmented Dickey-Fuller (ADF) Test
    $$\Delta \epsilon_t = \gamma \epsilon_{t-1} + \sum_{i=1}^p \delta_i \Delta \epsilon_{t-i} + u_t$$
3) Hypothesis H0 : gamma = 0 (unit root exists, pair does not mean revert)
              H1 : gamma < 0 (stationary - mean revertin spread)
4) Need to test the statistic to be more engativce than critical value to confirm cointegration


Rolling Z-Score NormalizationConcept: Normalizing the fluctuating spread relative to its historical rolling window (e.g., 30-day window):
$$Z_t = \frac{\text{Spread}_t - \mu_w}{\sigma_w}$$Trading Thresholds:If $Z_t > +2.0$: KO is overpriced relative to PEP $\rightarrow$ Short KO, Long PEP.If $Z_t < -2.0$: KO is underpriced relative to PEP $\rightarrow$ Long KO, Short PEP.If $\vert{}Z_t\vert{} \le 0.5$: Spread has converged to its equilibrium mean $\rightarrow$ Flatten positions (EXIT).
